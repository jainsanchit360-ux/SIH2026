**Project:** ResQtech — AI-Based Early Warning & Landslide Risk Monitoring System (NER India)

## Objectives:
- Load and inspect Geological Survey of India (GSI) historical landslide dataset extracted from PDF inventory (`data/processed/gsi_landslides_ner.csv` & `.geojson`).
- Perform spatial sanity checks, CRS verification (`EPSG:4326`), and state-wise record count breakdown across the 8 NER states.
- Inspect data quality metrics, coordinate coverage, missing value percentages, and landslide material/movement types.

```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import seaborn as sns
import json
from pathlib import Path

PROCESSED_DIR = Path("../data/processed")
REPORTS_DIR = Path("../reports")

# Load processed NER GSI GeoJSON dataset
gdf = gpd.read_file(PROCESSED_DIR / "gsi_landslides_ner.geojson")
print(f"Loaded {len(gdf)} NER landslide points. CRS: {gdf.crs}")
gdf.head()
```

## State-Wise Record Distribution (NER)

```python
state_counts = gdf["state_clean"].value_counts()
print("Landslides per State:")
print(state_counts)

plt.figure(figsize=(10, 5))
sns.barplot(x=state_counts.index, y=state_counts.values, hue=state_counts.index, palette="viridis", legend=False)
plt.title("NER Historical Landslide Records by State")
plt.ylabel("Count")
plt.xlabel("State")
plt.xticks(rotation=30)
plt.show()
```
